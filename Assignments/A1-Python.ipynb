### Exercise II.1
Write a **generator** ```page_provider``` which provides the pages of the PDF as strings.
The generator is supposed to expect the path to the PDF ```paper_path```
as an argument and, when, for example, used in a loop, provide the next page of
the document.

Fill in the cell below:

````python
# Exercise II.1
# Goal: create a generator called page_provider.
# A generator is a function that uses yield instead of return.
# In Notebook 2, squared_sequence yielded one number at a time.
# Here we use the same idea, but yield one PDF page text at a time.
def page_provider(paper_path):
    # Open the PDF file.
    # fitz.open(...) gives us a document object that we can loop over.
    doc = fitz.open(paper_path)

    # Go through the document page by page.
    # On each loop step, page is one PDF page object.
    for page in doc:
        # Extract the visible text from the current PDF page.
        page_text = page.get_text()

        # yield sends this page text back to the user of the generator.
        # The function pauses here until the next page is requested.
        yield page_text

    # Close the document after all pages were provided.
    doc.close()


# Small check: use the generator in a for-loop.
# This prints the pages one after another, just like the example above.
for page_text in page_provider(paper_path):
    print(page_text)
````

### Exercise II.2
Extract the abstract and conclusion sections from the PDF into a string variable.

For this, assign the generator ```page_provider``` to a variable e.g. ```get_page```. Loop over the pages and extract the sections on 'ABSTRACT'
and 'CONCLUSION' and store the text in one string variable ```result_text```.

Ideas on strings provided in [Notebook 2](../Notebooks/2-Python_Concepts.ipynb) might help you with that.

Fill in the cell below:

````python
# Exercise II.2
# Goal: collect the ABSTRACT and CONCLUSION sections in one string.

# Create the generator object.
# No pages are read yet; this only prepares the generator.
get_page = page_provider(paper_path)

# Start with an empty string.
# We will add the abstract and conclusion to this variable.
result_text = ""

# Loop over all page strings from the generator.
for page_text in get_page:

    # Check whether the current page contains the abstract heading.
    if "ABSTRACT" in page_text:
        # split("ABSTRACT") cuts the text into the part before and after ABSTRACT.
        # [1] keeps the part after ABSTRACT.
        text_after_abstract = page_text.split("ABSTRACT")[1]

        # The abstract ends before the INTRODUCTION heading.
        # Splitting again lets us keep only the abstract paragraph.
        abstract_text = text_after_abstract.split("INTRODUCTION")[0]

        # Add a label and the extracted text to the final result string.
        result_text += "ABSTRACT\n"
        result_text += abstract_text.strip()
        result_text += "\n\n"

    # Check whether the current page contains the conclusion heading.
    if "CONCLUSION" in page_text:
        # Keep the text after the CONCLUSION heading.
        text_after_conclusion = page_text.split("CONCLUSION")[1]

        # In this paper, the conclusion ends before REFERENCES.
        conclusion_text = text_after_conclusion.split("REFERENCES")[0]

        # Add a label and the extracted conclusion to the same result string.
        result_text += "CONCLUSION\n"
        result_text += conclusion_text.strip()
````

````python
print(result_text)
````

## Part III: Read a table from a PDF

In the third part, we tackle the slightly more challenging task of extracting 
tables from PDFs. We use the exemplary table in ```data/A1_table_example.pdf``` for this.

The following cell provide a wrapper function ```get_pandas_df``` to load the first table 
occurring on a specified page into a Pandas DataFrame. You can ignore the arguments
```search_strategy```and ```search_region```for now. Read up on the used ```PyMuPDF``` function 
in the [find_tables documentation](https://pymupdf.readthedocs.io/en/latest/page.html#Page.find_tables). 

````python
import fitz
import pandas as pd
import numpy as np


def get_pandas_df(page, search_strategy="text", search_region=None):
    tabs = page.find_tables(strategy=search_strategy, clip=search_region)

    tab = tabs[0]
    df = tab.to_pandas()

    df = df.convert_dtypes()

    return df
````

````python
paper_path = "data/A1_table_example.pdf"

doc = fitz.open(paper_path)

print(f"The paper '{paper_path}' has {len(doc)} pages.")

table_df = get_pandas_df(doc[0])
````

### Exercise III.1

As you probably have noticed ```table_df``` requires some addtional processing
before being properly used as DataFrame.

Thus, write a **decorator** ```nicer_df``` which can be used to conveniently extend
the functionality of our ```get_pandas_df```. In particular, please do not change the ```get_pandas_df``` function, but define a decorator as discussed in [Notebook 2](../Notebooks/2-Python_Concepts.ipynb).

Complete the cell below and perform the following steps:

* Drop all rows where entries with only empty strings ```''``` make up at least half of the entries in that row 
* Convert columns with numeric values to floats or integers, respectively
* for each numeric row, provide as the last two rows the mean and standard deviation

Hints: Your solution could include the following parts

*  ```== ''```
*  ```.sum(axis=1)```
*  ```~``` operator for negation, i.e. ```~np.array([True, True, False])``` becomes ```array([False, False,  True])```
*  ```pd.to_numeric(..., errors='ignore')```
*  ```df.loc[row_index, column_name]```

Note that your own solution might not require all of these.

````python
# Exercise III.1
# Goal: write a decorator that improves the DataFrame returned by get_pandas_df.
# A decorator is a function that receives another function as input.
def nicer_df(func):

    # wrapper will replace the original function after decoration.
    # *args and **kwargs allow us to pass on all arguments unchanged.
    def wrapper(*args, **kwargs):

        # First run the original function.
        # For this assignment, func is get_pandas_df.
        df = func(*args, **kwargs)

        # Make a copy before cleaning.
        # This keeps the original returned DataFrame separate from our edits.
        df = df.copy()

        # Part 1: remove rows that are mostly empty strings.
        # (df == "") creates a True/False table.
        # True means that this entry is an empty string.
        empty_string_table = df == ""

        # Sum the True values in each row.
        # In Python, True behaves like 1 and False behaves like 0.
        empty_entries_per_row = empty_string_table.sum(axis=1)

        # A row should be dropped if at least half of its entries are empty.
        half_number_of_columns = len(df.columns) / 2
        rows_to_drop = empty_entries_per_row >= half_number_of_columns

        # Keep the opposite of rows_to_drop.
        # The ~ operator changes True to False and False to True.
        df = df[~rows_to_drop]

        # Part 2: convert columns to numbers where possible.
        # Some columns contain text, for example train/test or sex.
        # Those columns cannot be converted and are left unchanged.
        for column in df.columns:
            try:
                df[column] = pd.to_numeric(df[column])
            except ValueError:
                # If conversion fails, keep this column as it is.
                pass

        # Part 3: add mean and standard deviation for numeric columns.
        # select_dtypes keeps only columns with numeric values.
        numeric_columns = df.select_dtypes(include=[np.number]).columns

        # Add the mean values in a new last row called "Mean".
        df.loc["Mean", numeric_columns] = df[numeric_columns].mean()

        # Add the standard deviations in another last row called "Std".
        df.loc["Std", numeric_columns] = df[numeric_columns].std()

        # Return the cleaned and extended DataFrame.
        return df

    # Return the wrapper function, not the result of wrapper.
    return wrapper
````

If you were successful with performing all the outlined steps,
you can *decorate* the ```get_pandas_df``` like this:

````python
@nicer_df
def get_pandas_df(page, search_strategy="text", search_region=None):
    tabs = page.find_tables(strategy=search_strategy, clip=search_region)

    tab = tabs[0]
    df = tab.to_pandas()

    df = df.convert_dtypes()

    return df
````

Re-run the table extraction with the extended functionality:

````python
import fitz
import pandas as pd
import numpy as np

paper_path = "data/A1_table_example.pdf"

doc = fitz.open(paper_path)

print(f"The paper '{paper_path}' has {len(doc)} pages.")

nicer_table_df = get_pandas_df(doc[0])
````
